### 5.2 Attrition Distribution
`value_counts()` shows how many employees stayed and how many left.

| Attrition | Employees | Share |
|-----------|-----------|-------|
| No (stayed) | 1,233 | ~83.9% |
| Yes (left) | 237 | ~16.1% |

**Insight:** About **1 in 6 employees (~16%) has left** the company. The data is **imbalanced**, with far more "No" than "Yes", which should be kept in mind when building any prediction model later.

```python
df['Attrition'].value_counts()
```
